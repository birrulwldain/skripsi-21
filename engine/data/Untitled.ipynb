```python
import sqlite3
import os

def export_spectrum_to_asc(db_file_path, output_dir="exported_spectrum_data"):
    """
    Mengekspor data spektrum dari tabel 'spectrum_data' ke file .asc terpisah.
    Setiap file .asc mewakili satu spektrum unik (berdasarkan sample_name dan iteration).
    Format file .asc akan menjadi dua kolom: Wavelength dan Intensity.

    Args:
        db_file_path (str): Jalur lengkap ke file database .db.
        output_dir (str): Nama direktori tempat file .asc akan disimpan.
                          Jika direktori tidak ada, akan dibuat.
    """
    conn = None
    try:
        conn = sqlite3.connect(db_file_path)
        cursor = conn.cursor()

        # Buat direktori output jika belum ada
        if not os.path.exists(output_dir):
            os.makedirs(output_dir)
            print(f"Direktori output '{output_dir}/' telah dibuat.")

        print(f"\n--- Mengekspor data dari '{db_file_path}' ke file .asc ---")

        # Dapatkan daftar unik dari sample_name dan iteration
        cursor.execute("SELECT DISTINCT sample_name, iteration FROM spectrum_data ORDER BY sample_name, iteration;")
        unique_spectrums = cursor.fetchall()

        if not unique_spectrums:
            print("Tidak ada data spektrum unik (sample_name, iteration) ditemukan.")
            return

        exported_count = 0
        for sample_name, iteration in unique_spectrums:
            # Ambil data wavelength dan intensity untuk spektrum spesifik ini
            cursor.execute(
                "SELECT wavelength, intensity FROM spectrum_data WHERE sample_name = ? AND iteration = ? ORDER BY wavelength;",
                (sample_name, iteration)
            )
            spectrum_data = cursor.fetchall()

            if spectrum_data:
                # Buat nama file: contoh_sampel_iterasi_1.asc
                # Mengganti karakter yang tidak valid dalam nama file
                safe_sample_name = "".join(c for c in sample_name if c.isalnum() or c in (' ', '_', '-')).strip()
                file_name = f"{safe_sample_name.replace(' ', '_')}_iteration_{iteration}.asc"
                file_path = os.path.join(output_dir, file_name)

                with open(file_path, 'w') as f:
                    # Tulis header (opsional, tapi disarankan)
                    f.write("# Wavelength (nm)\tIntensity\n")
                    for wavelength, intensity in spectrum_data:
                        f.write(f"{wavelength}\t{intensity}\n")
                exported_count += 1
                print(f"  Berhasil mengekspor: {file_name}")
            else:
                print(f"  Peringatan: Tidak ada data untuk Sample: '{sample_name}', Iterasi: {iteration}")

        print(f"\nSelesai. Total {exported_count} file .asc diekspor ke direktori '{output_dir}/'.")

    except sqlite3.Error as e:
        print(f"Terjadi kesalahan SQLite: {e}")
    except Exception as e:
        print(f"Terjadi kesalahan tak terduga: {e}")
    finally:
        if conn:
            conn.close()
            print("Koneksi database ditutup.")

# --- Contoh Penggunaan ---
if __name__ == "__main__":
    # Ganti dengan jalur file .db Anda yang sebenarnya
    db_file = '/Volumes/Private-C/Private-C-backup/skripsi-21/engine/lll.db'
    
    # Direktori tempat file .asc akan disimpan. Ini akan dibuat di lokasi yang sama dengan skrip Python.
    output_directory = "exported_spectrum_data" 

    export_spectrum_to_asc(db_file, output_directory)
```
